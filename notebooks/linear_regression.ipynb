```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
df= pd.read_csv('../data/parkinsons_updrs.data')
```

```python
df.head()
```

```python
df=df.drop(columns=['subject#'])
```

```python
df.head()
```

```python
X=df.drop(columns=['motor_UPDRS', 'total_UPDRS'])
y=df['total_UPDRS']
```

```python
X.head()
```

```python
y.head()
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X,y, test_size=0.20, random_state=42)
```

```python
X.shape, y.shape
```

```python
X_train.shape, X_test.shape, y_train.shape, y_test.shape
```

```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
```

```python
X_train=scaler.fit_transform(X_train)
X_test=scaler.transform(X_test)
```

```python
X_train
```

```python
from sklearn.linear_model import LinearRegression
regression=LinearRegression()
regression.fit(X_train, y_train)
```

```python
y_pred_test=regression.predict(X_test)
```

```python
comparison = pd.DataFrame({
    'Actual'   : y_test.values[:10],
    'Predicted': y_pred_test[:10].round(2)
})
print('Sample Predictions (first 10):')
print(comparison)
```

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error
mse=mean_squared_error(y_test, y_pred_test)
mae=mean_absolute_error(y_test, y_pred_test)
rmse=np.sqrt(mse)
print("MSE:", mse)
print("MAE:", mae)
print("RMSE:", rmse)
```

```python
from sklearn.metrics import r2_score
score=r2_score(y_test, y_pred_test)
print("R2 Score:", score)
```

```python
residuals = y_test - y_pred_test

plt.figure(figsize=(10, 5))
sns.histplot(residuals, bins=30, kde=True, color='tomato')
plt.title('Residual Distribution — Linear Regression')
plt.xlabel('Residual (Actual - Predicted)')
plt.ylabel('Count')
plt.axvline(0, color='black', linestyle='--', linewidth=1.5)
plt.show()
```

---
## 📝 Conclusion

Linear Regression gave an R² of only 0.15, meaning it could explain
just 15% of the data. This is expected since voice features have complex
nonlinear relationships that a simple straight line cannot capture.
Moving to Random Forest Regressor for better results.
